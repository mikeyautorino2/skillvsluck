```python
import numpy as np
import pandas as pd
```

```python
def simulate_coin_flip_dataset(in_path="../csv/us_combined_data.csv", p=0.5, out_dir="../csv/coin_flip_simulations/",seed=43) -> str:
    
    if not (0.0 <= p <= 1.0):
        raise ValueError("p must be in [0, 1].")

    df = pd.read_csv(in_path)
    n = len(df)
    rng = np.random.default_rng(seed)

    # Bernoulli(p): 1 = home win, 0 = away win
    home_win = rng.binomial(1, p, size=n)

    # result: +1 (home) or -1 (away)
    df["result"] = np.where(home_win == 1, 1, -1).astype(int)

    # Set scores so winner=1, loser=0
    df["score1"] = np.where(df["result"] == 1, 1, 0).astype(int) 
    df["score2"] = np.where(df["result"] == -1, 1, 0).astype(int) 

    # filename suffix for p (e.g., 0.5, 0.7, 0.05)
    p_str = f"{p:.3f}".rstrip("0").rstrip(".") or "0"
    out_path = f"{out_dir}/us_combined_data_coin_flip_{p_str}_s{seed}.csv"

    df.to_csv(out_path, index=False)
    return out_path
```

```python
def simulate_coin_flip_home_bias(
    in_path="../../data/us_leagues/csv/game_by_game/actual/us_combined_data.csv",
    out_path="../../data/us_leagues/csv/game_by_game/pure_luck/coin_flip_home_bias_us.csv",
    seed=43,
) -> str:
    """
    Home win percentage from box plots:
      MLB: 0.5380  (0.5 + 0.0380)
      NBA: 0.6047  (0.5 + 0.1047)
      NFL: 0.5725  (0.5 + 0.0725)
    """
    HOME_P = {"MLB": 0.5380, "NBA": 0.6047, "NFL": 0.5725}

    df = pd.read_csv(in_path)
    if "league" not in df.columns:
        raise ValueError("CSV does not have a 'league' column.")

    # Per row home win probabilities (default 0.5)
    home_win_prob = df["league"].map(HOME_P).fillna(0.5).to_numpy(float)
    home_win_prob = np.clip(home_win_prob, 0.0, 1.0)

    rng = np.random.default_rng(seed)
    home_win = rng.binomial(1, home_win_prob, size=len(df))  # 1=home wins, 0=away

    df["result"] = np.where(home_win == 1, 1, -1).astype(int)
    df["score1"] = (df["result"] == 1).astype(int)
    df["score2"] = (df["result"] == -1).astype(int)

    df.to_csv(out_path, index=False)
    print(f"Wrote biased coin flip sim to {out_path} (rows: {len(df):,})")
    return out_path
```
