```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_mlb = pd.read_csv('../csv/mlb_data.csv')
df_nba = pd.read_csv('../csv/nba_data.csv')
df_nfl = pd.read_csv('../csv/nfl_data.csv')
```

```python
mlb_score1 = df_mlb['score1']
mlb_score2 = df_mlb['score2']
nba_score1 = df_nba['score1']
nba_score2 = df_nba['score2']
nfl_score1 = df_nfl['score1']
nfl_score2 = df_nfl['score2']
```

```python
def calculate_statistics(series_list, titles):
    """
    series_list : list[pd.Series]
    titles      : list[str]
    Returns a DataFrame with descriptive statistics.
    """
    rows = []
    for title, s in zip(titles, series_list):
        s = s.dropna()
        rows.append({
            "Series"   : title,
            "#obs"     : s.count(),
            "mean"     : round(s.mean(), 3),
            "mode"     : round(s.mode().iloc[0], 3) if not s.mode().empty else np.nan,
            "std"      : round(s.std(), 3),
            "variance" : round(s.var(), 3),
            "skewness" : round(s.skew(), 3),
            "kurtosis" : round(s.kurtosis(), 3),
            "min"      : round(s.min(), 3),
            "Q1"       : round(s.quantile(0.25), 3),
            "median(Q2)": round(s.median(), 3),
            "Q3"       : round(s.quantile(0.75), 3),
            "max"      : round(s.max(), 3),
            "IQR"      : round(s.quantile(0.75) - s.quantile(0.25), 3)
        })

    return pd.DataFrame(rows)

calculate_statistics([mlb_score1, mlb_score2, nba_score1, nba_score2, nfl_score1, nfl_score2], ["MLB Score1", "MLB Score2", "NBA Score1", "NBA Score2", "NFL Score1", "NFL Score2"])
```

```python
mlb_results = df_mlb['result']
nba_results = df_nba['result']
nfl_results = df_nfl['result']

series_dict = {
    "MLB": mlb_results,
    "NBA": nba_results,
    "NFL": nfl_results
}

rows = []
for name, s in series_dict.items():
    counts = s.value_counts().reindex([3, 1, 0], fill_value=0)
    total = counts.sum()
    rows.append({
        "Series"         : name,
        "WIN_count"      : counts[3],
        "WIN_pct(%)"     : round(counts[3] / total * 100, 2),
        "LOSS_count"     : counts[0],
        "LOSS_pct(%)"    : round(counts[0] / total * 100, 2),
        "TIE_count"      : counts[1],
        "TIE_pct(%)"     : round(counts[1] / total * 100, 2),
        "Total_games"    : total
    })

summary_df = pd.DataFrame(rows)
summary_df
```

```python
def plot_histogram(data, color, title):
    bins = np.arange(data.min() - 0.5, data.max() + 1.5, 1)
    plt.hist(data, bins=bins, color=color)
    plt.title(title)
    plt.xlabel('Score')
    plt.ylabel('Frequency')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

plot_histogram(mlb_score1, 'blue', 'Histogram of score1 in MLB Games')
plot_histogram(mlb_score2, 'orange', 'Histogram of score2 in MLB Games')
plot_histogram(nba_score1, 'green', 'Histogram of score1 in NBA Games')
plot_histogram(nba_score2, 'red', 'Histogram of score2 in NBA Games')
plot_histogram(nfl_score1, 'skyblue', 'Histogram of score1 in NFL Games')
plot_histogram(nfl_score2, 'pink', 'Histogram of score2 in NFL Games')
```
