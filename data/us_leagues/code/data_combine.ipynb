```python
import numpy as np
import pandas as pd
```

```python
df_mlb = pd.read_csv('../csv/mlb_data.csv')
df_nba = pd.read_csv('../csv/nba_data.csv')
df_nfl = pd.read_csv('../csv/nfl_data.csv')
```

```python
df_mlb.insert(2, 'league', 'MLB')
df_nba.insert(2, 'league', 'NBA')
df_nfl.insert(2, 'league', 'NFL')
```

```python
df_list = [df_mlb, df_nba, df_nfl]
df_all = pd.concat(df_list, ignore_index=True)
```

```python
df_all.drop("home_away", axis=1, inplace=True)
```

```python
df_all['result'] = df_all['result'].map({3:1, 1:0, 0:-1})
```

```python
df_all.to_csv('../csv/us_combined_data.csv', index=False)
```
