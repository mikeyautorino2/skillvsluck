```python
from google.colab import drive
drive.mount('/content/drive')
import numpy as np
import pandas as pd
```

```python
df_list = []
for year in range(1980, 2025):
    df = pd.read_csv(f"drive/My Drive/IML/mlb_dataset/raw_data/gl{year}.txt", delimiter=",", header=None)
    df_list.append(df)
df_all = pd.concat(df_list)
```

```python
df_filtered = df_all.iloc[:,[0, 6, 3, 10, 9]]
df_filtered.rename(columns={0: 'date', 6: 'team1', 3: 'team2', 10: 'score1', 9: 'score2'}, inplace=True)
df_filtered
```

```python
df_filtered['date'] = pd.to_datetime(df_filtered['date'], format='%Y%m%d')
df_filtered.insert(0, 'season', df_filtered['date'].dt.year)
```

```python
conditions = [df_filtered['score1'] > df_filtered['score2'], df_filtered['score1'] < df_filtered['score2'], df_filtered['score1'] == df_filtered['score2']]
df_filtered.insert(4, 'result', np.select( conditions, ['W', 'L', 'T'], default='ERROR'))
```

```python
df_filtered.insert(7, 'home_flag', 'HOME')
```

```python
df_filtered.to_csv('drive/My Drive/IML/mlb_dataset/mlb_data_cleaned.csv', index=False)
```
