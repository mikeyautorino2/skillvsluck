```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
from pathlib import Path
```

```python
BASE = "/content/drive/MyDrive/Sophomore Year/IML_Fall2025_SkillVersusLuck/Simulated Leagues"
IN_FP  = f"{BASE}/la_liga_simulated_matches.csv"
OUT_FP = f"{BASE}/la_liga_simulated_standings_all_seasons.csv"

df = pd.read_csv(IN_FP)
```

```python
res = df["simulated_home_team_result"].astype(int)

df["home_pts"] = (res.eq(1)*3 + res.eq(0)*1).astype(int)
df["away_pts"] = (res.eq(-1)*3 + res.eq(0)*1).astype(int)
```

```python
df["home_w"] = res.eq(1).astype(int)
df["home_d"] = res.eq(0).astype(int)
df["home_l"] = res.eq(-1).astype(int)

df["away_w"] = res.eq(-1).astype(int)
df["away_d"] = res.eq(0).astype(int)
df["away_l"] = res.eq(1).astype(int)
```

```python
home = pd.DataFrame({
    "season": df["season"].astype(int),
    "team":   df["team1"],
    "points": df["home_pts"],
    "wins":   df["home_w"],
    "draws":  df["home_d"],
    "losses": df["home_l"],
})
away = pd.DataFrame({
    "season": df["season"].astype(int),
    "team":   df["team2"],
    "points": df["away_pts"],
    "wins":   df["away_w"],
    "draws":  df["away_d"],
    "losses": df["away_l"],
})
```

```python
long = pd.concat([home, away], ignore_index=True)

stand = (long.groupby(["season","team"], as_index=False)
             .sum(numeric_only=True))
```

```python
stand = stand.sort_values(["season","points","team"],
                          ascending=[True, False, True],
                          kind="mergesort")

stand["rank"] = stand.groupby("season").cumcount() + 1
```

```python
stand = stand[["season","team","points","wins","draws","losses","rank"]]
stand.to_csv(OUT_FP, index=False)
print(f"Saved: {OUT_FP}")
```
