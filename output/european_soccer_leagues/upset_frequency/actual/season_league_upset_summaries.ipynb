```python
import pandas as pd
from pathlib import Path
```

```python
BASE = Path(".")

season_files = list(BASE.glob("*_seasonal_upset_frequency.csv"))
season_files
```

```python
dfs = []

for path in season_files:
    df = pd.read_csv(path)

    df_small = df[["league", "season", "total_predictions", "total_upsets", "upset_frequency"]].copy()
    df_small = df_small.rename(columns={"total_predictions": "total_matches"})

    dfs.append(df_small)

season_summary = pd.concat(dfs, ignore_index=True)

# sort
season_summary = season_summary.sort_values(["league", "season"])

# output
season_out = BASE / "season_upset_frequency_summary.csv"
season_summary.to_csv(season_out, index=False)
season_out
```

```python
league_summary = (
    season_summary
    .groupby("league", as_index=False)
    .agg(
        total_matches=("total_matches", "sum"),
        total_upsets=("total_upsets", "sum"),
    )
)

league_summary["upset_frequency"] = league_summary["total_upsets"] / league_summary["total_matches"]

league_out = BASE / "league_upset_frequency_summary.csv"
league_summary.to_csv(league_out, index=False)
league_out
```
